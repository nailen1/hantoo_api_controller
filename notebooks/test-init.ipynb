API 연결 및 주요 기능 동작 확인을 위한 노트북입니다.

**사전 준비:**
1. `.env.example`을 복사하여 `.env` 파일 생성
2. 한국투자증권에서 발급받은 APP_KEY, APP_SECRET, 계좌번호 입력
3. `pip install -r requirements.txt` 실행

```python
import sys
sys.path.append('..')

from hantoo_api_controller import HantooSession

session = HantooSession()
```

## 1. 계좌 잔고 조회

```python
holdings, summary = session.get_info_account(option_kor_cols=True)

print("보유 종목:")
display(holdings)
print("\n계좌 요약:")
display(summary)
```

## 2. 현재가 조회

```python
symbol = "005930"
price = session.current_price_value(symbol)
print(f"{symbol} 현재가: {price:,}원")
```

## 3. 매수 주문 바디 구성 (주문 전송 없음)

```python
body = session.build_buy_order_body(symbol, quantity=1, order_type="limit", price=price)
body
```

## 4. 금액 기준 매수 주문 바디 구성 (주문 전송 없음)

```python
body = session.build_buy_order_body_by_amount(symbol, amount=1_000_000)
print(f"단가: {body['ORD_UNPR']}원 × {body['ORD_QTY']}주")
body
```

## 5. 매도 주문 바디 구성 (금액 기준, 주문 전송 없음)

```python
body = session.build_sell_order_body_by_amount(symbol, amount=1_000_000)
print(f"단가: {body['ORD_UNPR']}원 × {body['ORD_QTY']}주")
body
```

## 6. 주문 체결 내역 조회

```python
orders, order_summary = session.get_order_history(option_kor_cols=True)

print(f"오늘 주문 건수: {len(orders)}건")
if not orders.empty:
    display(orders)
else:
    print("오늘 주문 내역이 없습니다.")
```
